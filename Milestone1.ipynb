TOPIC 3:DATA TRANSFORMATION

```python
df_transformed=df.copy()
print("Original dataset shape:",df.shape)
print("transformed dataset shape:",df_transformed.shape)
```

```python
df_transformed.columns=(df_transformed.columns.str.strip()
                        .str.lower()
                        .str.replace(" ",
                                     "_")
                        )
print("Updated column names:")
print(df_transformed.columns.tolist())
```

```python
selected_columns=["artist","track","album_type","danceability","energy","loudness"]
display(df_transformed[selected_columns].head())
```

```python
numeric_columns=["danceability","energy","loudness"]
for column in numeric_columns:df_transformed[column]=pd.to_numeric(df_transformed[column],errors="coerce")
print(df_transformed[numeric_columns].dtypes)
```

```python
df_transformed["loudness_magnitude"]=np.abs(df_transformed["loudness"])
display(df_transformed[["loudness","loudness_magnitude"]].head(10))
```

topic 4:handling missing values

```python
print("Missing values in each column:")
print(df_transformed.isnull().sum())
```

```python
total_missing=df_transformed.isnull().sum().sum()
print("total missing values:",total_missing)
```

```python
missing_rows=df_transformed[df_transformed.isnull().any(axis=1)]
print("number of rows containing missing values:",len(missing_rows))
display(missing_rows.head(10))
```

```python
numerical_columns = df_transformed.select_dtypes(
    include=np.number
).columns

for column in numerical_columns:
    if df_transformed[column].isnull().any():
        mean_value = df_transformed[column].mean()
        df_transformed[column] = df_transformed[column].fillna(mean_value)

print("Missing numerical values handled using mean imputation.")
```

```python
categorical_columns = df_transformed.select_dtypes(
    include=["object", "category", "string"]
).columns

for column in categorical_columns:
    if df_transformed[column].isnull().any():
        mode_values = df_transformed[column].mode()

        if not mode_values.empty:
            df_transformed[column] = df_transformed[column].fillna(
                mode_values[0]
            )

print("Missing categorical values handled where possible.")
```

```python
print("missing values after handling:")
print(df_transformed.isnull().sum())
print("total missing values remaining :",
df_transformed.isnull().sum().sum()
)
```

```python
print("Total missing values:", df_transformed.isnull().sum().sum())

print("\nColumns containing missing values:")
print(df_transformed.columns[df_transformed.isnull().any()].tolist())
```

in this spotify dataset it showed zero missing values remaining .therefore no additional implementation is required for the current dataset.

TOPIC 4:INSIGHT GENERATION

```python
correlation_matrix=df_transformed.select_dtypes(include=np.number).corr()
print("correlation matrix:")
display(correlation_matrix.round(2))
```

```python
plt.figure(figsize=(14,10))
correlation_matrix=df_transformed.select_dtypes(include=np.number).corr()
sns.heatmap(correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5)
plt.title("correlation heatmap of spotify features")
plt.tight_layout()
plt.show()
```

```python
correlation_matrix = df_transformed.select_dtypes(
    include=np.number
).corr()

correlation_pairs = correlation_matrix.unstack()

correlation_pairs = correlation_pairs[
    correlation_pairs.index.get_level_values(0)
    != correlation_pairs.index.get_level_values(1)
]

correlation_pairs = correlation_pairs.drop_duplicates()

print("Top 5 positive correlations:")
display(correlation_pairs.sort_values(ascending=False).head(5))

print("Top 5 negative correlations:")
display(correlation_pairs.sort_values(ascending=True).head(5))
```

```python
feature_analysis=df_transformed[["danceability","energy","valence","acousticness"]].mean()
print("average spotify audio features:")
display(feature_analysis.round(3))
```
